```python

from nir.tests.evaluator import run_generation_tests

model_config = ModelConfig(
    model_name="hf.co/VlSav/Vikhr-Nemo-12B-Instruct-R-21-09-24-Q4_K_M-GGUF:latest", 
    temperature=0
)
eval_llm = manager.create_chat_model(name="test_model", option="ollama", config=model_config)

raw_results = {}

for ds_name, ds_data in DATASETS.items():
    raw_results[ds_name] = {}
    for p_name, pipeline in pipelines.items():
        df = run_generation_tests(
            dataset=ds_data,
            dataset_name=ds_name,
            pipeline=pipeline,
            evaluation_llm=eval_llm,
            embedding_model=embed_model
        )
        raw_results[ds_name][p_name] = df

print("Все тесты завершены. JSON-файлы сохранены в assets/outputs/test_results/")
```

```python
import os
import json
import pandas as pd
import numpy as np

base_output_dir = "assets/outputs/test_results"
raw_results = {}

# 1. Сканируем папки пайплайнов и загружаем JSON
for p_name in os.listdir(base_output_dir):
    p_dir = os.path.join(base_output_dir, p_name)
    if not os.path.isdir(p_dir):
        continue

    for ds_file in os.listdir(p_dir):
        if not ds_file.endswith('.json'):
            continue

        ds_name = os.path.splitext(ds_file)[0]  # Имя датасета из имени файла
        filepath = os.path.join(p_dir, ds_file)

        with open(filepath, 'r', encoding='utf-8') as f:
            data = json.load(f)

        # 2. Извлекаем метрики по каждому запросу
        records = []
        for item in data:
            cat = item.get("category", "unknown")
            metrics = item.get("metrics", {})
            
            # Оставляем только числовые поля (исключаем дублирующий 'category' внутри metrics)
            clean_metrics = {}
            for k, v in metrics.items():
                if k == "category": 
                    continue
                # На случай, если NaN сохранён как строка "NaN"
                if isinstance(v, str) and v.lower() == 'nan':
                    v = np.nan
                if isinstance(v, (int, float, np.number)):
                    clean_metrics[k] = v
                    
            clean_metrics["category"] = cat
            records.append(clean_metrics)

        if not records:
            continue

        df_metrics = pd.DataFrame(records)
        
        # 3. Считаем средние значения по каждой категории
        # numeric_only=True автоматически игнорирует NaN при усреднении
        cat_df = df_metrics.groupby("category").mean(numeric_only=True)
        
        # 4. Добавляем строку OVERALL (среднее по всем категориям и метрикам)
        cat_df.loc["OVERALL"] = df_metrics.mean(numeric_only=True)
        
        # 5. Сохраняем в структуру raw_results[ds_name][p_name]
        if ds_name not in raw_results:
            raw_results[ds_name] = {}
        raw_results[ds_name][p_name] = cat_df

print("✅ JSON-файлы загружены и агрегированы в raw_results")
```

```python
CATEGORIES = ["quest", "dialogue", "character description", "location description", "item description", "OVERALL"]
ds_category_tables = {ds: {} for ds in DATASETS}

print("📊 ТАБЛИЦЫ ПО ДАТАСЕТАМ И КАТЕГОРИЯМ")
for ds_name, p_dfs in raw_results.items():
    for cat in CATEGORIES:
        rows = {}
        for p_name, df in p_dfs.items():
            if cat in df.index:
                rows[p_name] = df.loc[cat]
        if rows:
            table = pd.DataFrame(rows).T 
            ds_category_tables[ds_name][cat] = table
            
            display(
                table.style
                .set_caption(f"📁 {ds_name.upper()} • 🏷️ {cat}")
                .format("{:.4f}")
                .highlight_max(axis=0, color="lightgreen")
                .highlight_min(axis=0, color="lightcoral")
            )
```

```python
print("\n🏆 ФИНАЛЬНЫЕ СРЕДНИЕ ТАБЛИЦЫ (6 штук)")
final_tables = {}

for cat in CATEGORIES:
    dfs_to_avg = [ds_category_tables[ds][cat] for ds in DATASETS if cat in ds_category_tables[ds]]
    if dfs_to_avg:
        avg_df = pd.concat(dfs_to_avg).groupby(level=0).mean()
        final_tables[cat] = avg_df
        display(
            avg_df.style
            .set_caption(f"📈 Среднее по всем датасетам • {cat}")
            .format("{:.4f}")
            .highlight_max(axis=0, color="lightgreen")
            .highlight_min(axis=0, color="lightcoral")
        )
```
